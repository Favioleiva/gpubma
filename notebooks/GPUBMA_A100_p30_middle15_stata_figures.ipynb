# GPUBMA — exact exhaustive 2^30 BMA on `panel_30_center15` (Google Colab, NVIDIA A100)

[![Open in Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/Favioleiva/gpubma/blob/main/notebooks/GPUBMA_A100_p30_middle15_stata_figures.ipynb)

Runs the **validated exact GPUBMA enumerator** (streamed direct batching,
float64 only, deterministic reductions, SHA-256-gated checkpoint/resume —
`docs/ADR_0001_GPU_ENUMERATOR.md`, `docs/FWL_BLOCK_FORMULATION.md`) on the
frozen **`panel_30_center15`** benchmark: p = 30 candidates
(**15 correlated true regressors x1–x15 + 15 structural-zero proxies
x16–x30**), n = 2,000, shrink (Stata-verified) convention, g = 2000,
beta-binomial(1, 1) model prior, **all 2^30 = 1,073,741,824 models exactly**
— no MCMC, no sampling, no pruning, no truncation, no early stopping.

**How to run**
1. Open this notebook in Colab (badge above, or `File → Open notebook →
   GitHub`).
2. Runtime → Change runtime type → **A100 GPU** (the completed canonical
   measurement used an A100-SXM4-80GB; do **not** run this on CPU).
3. Run the cells top to bottom: bootstrap → input validation → **mandatory
   smoke test** → (optionally mount Drive) → set `RUN_FULL_EXACT_P30 = True`
   in the config cell → run the expensive cell → validate → download the
   compact results ZIP.

**Cost warning.** The full run enumerates 1,073,741,824 models and consumes
Colab compute units. This completed `panel_30_center15` run measured
115.7 s for PASS1 and 896 s for exact PASS2 on an A100-SXM4-80GB. The
earlier ~95 s figure belongs to the old, easier `panel_30` sparse
benchmark and must not be attributed to this run.

**Storage modes — Google Drive is OPTIONAL and disabled by default.**
Everything the run needs (code, notebook, dataset, metadata) comes from the
public GitHub repository; no paid Drive plan, no Drive mount, no upload,
no GitHub credential, and no token are required.

- *Default, free mode* (`USE_GOOGLE_DRIVE = False`):
  GitHub inputs → Colab local runtime (`/content`) → manual checkpoint /
  result downloads. Colab local storage is **ephemeral**: checkpoints
  survive cell reruns within the same active runtime, but are lost if the
  runtime is destroyed — protect progress by downloading a checkpoint
  bundle (cell below) and re-uploading it after a reset.
- *Optional persistence mode* (`USE_GOOGLE_DRIVE = True`):
  checkpoints/results live on your Drive and a destroyed session resumes
  automatically after Run all.

**Scientific expectation.** The benchmark is *designed* so that proxies
compete with true variables (proxy–primary correlations 0.79–0.89).
Posterior mass may legitimately spread across observationally similar
models, and **the exact true model need not have the highest PMP** —
family-level recovery (true+proxy groups) is the scientifically meaningful
outcome. See `reports/panel_30_center15_dgp_validation.md`.

```python
# [VALIDATE INPUT] Load the canonical dataset + metadata from the pinned
# public clone and verify every documented property BEFORE any enumeration.
import hashlib, json, math, pathlib, time
import numpy as np
import pandas as pd

t_load0 = time.perf_counter()
DATA_PARQUET = pathlib.Path(REPO_DIR) / DATASET_REL
META_PATH = pathlib.Path(REPO_DIR) / METADATA_REL
META = json.loads(META_PATH.read_text())

PARQUET_SHA256 = hashlib.sha256(DATA_PARQUET.read_bytes()).hexdigest()
METADATA_SHA256 = hashlib.sha256(META_PATH.read_bytes()).hexdigest()
assert PARQUET_SHA256 == META['parquet_sha256'], (
    'Parquet SHA-256 mismatch — the clone does not contain the frozen '
    'canonical artifact')

DF = pd.read_parquet(DATA_PARQUET)
n = len(DF)
TRUE_VARS = [f'x{j}' for j in range(1, 16)]
PROXY_VARS = [f'x{j}' for j in range(16, 31)]
EXPECTED_COLS = ['individual_id', 'period', 'y'] + \
    [f'x{j}' for j in range(1, 31)] + ['w1', 'w2']

checks = []
def chk(name, ok, detail=''):
    checks.append((name, bool(ok), detail))
    if not ok:
        raise AssertionError(f'input validation FAILED: {name} ({detail})')

chk('generator seed = 20260724', META['seed'] == EXPECTED_SEED, META['seed'])
chk('n = 2,000', n == 2000, n)
chk('35 columns, expected names/order', list(DF.columns) == EXPECTED_COLS)
chk('30 candidate regressors', META['n_candidate_regressors'] == 30)
chk('15 true regressors', META['true_variables'] == TRUE_VARS)
chk('15 proxy regressors', META['proxy_variables'] == PROXY_VARS)
chk('dtypes (int32 ids, float64 data)',
    str(DF['individual_id'].dtype) == 'int32'
    and str(DF['period'].dtype) == 'int32'
    and all(str(DF[c].dtype) == 'float64' for c in EXPECTED_COLS[2:]))
chk('no missing values', not DF.isna().any().any())
chk('no non-finite values', np.isfinite(DF.to_numpy(np.float64)).all())

y = DF['y'].to_numpy(np.float64)
X30 = DF[[f'x{j}' for j in range(1, 31)]].to_numpy(np.float64)
A = np.column_stack([np.ones(n), DF[['w1', 'w2']].to_numpy(np.float64)])
Q, _ = np.linalg.qr(A)
D_true = np.column_stack([np.ones(n), DF[['w1', 'w2']].to_numpy(np.float64),
                          DF[TRUE_VARS].to_numpy(np.float64)])
Qt, _ = np.linalg.qr(D_true)
resid = y - Qt @ (Qt.T @ y)
yc = y - y.mean()
R2 = float(1.0 - (resid @ resid) / (yc @ yc))
SNR = R2 / (1.0 - R2)
chk('realized R^2 in [0.695, 0.705]', 0.695 <= R2 <= 0.705, f'{R2:.6f}')
chk('R^2 equals metadata value',
    abs(R2 - META['noise_calibration']['realized_r2']) < 1e-9)
chk('SNR in validated band [2.28, 2.39]', 2.28 <= SNR <= 2.39, f'{SNR:.4f}')
X_r = X30 - Q @ (Q.T @ X30)
chk('rank(X) = 30', np.linalg.matrix_rank(X30) == 30)
chk('rank(residualized X) = 30', np.linalg.matrix_rank(X_r) == 30)
eig = np.linalg.eigvalsh(X_r.T @ X_r)
COND = float(eig[-1] / eig[0])
chk('residualized Gram condition in validated range [50, 85]',
    50.0 <= COND <= 85.0, f'{COND:.2f}')

# validated shrink-convention inputs for the enumerator
Y_R = y - Q @ (Q.T @ y)
CONV = dict(df_resid=n - 1, tss_norm=float(yc @ yc), k_always=2)
G = float(max(n, P * P))          # = 2000 (benchmark rule max(n, p^2))
from gpubma.priors.model_priors import log_model_prior_function
LOG_PRIOR, PRIOR_DESC = log_model_prior_function(MODEL_PRIOR, P)
N_MODELS = 1 << P
TRUE_MASK = (1 << 15) - 1          # x1..x15 = bits 0..14
T_DATA_LOAD_S = time.perf_counter() - t_load0

print(f'{"check":55s} status')
for name, ok, detail in checks:
    print(f'{name:55s} {"PASS" if ok else "FAIL"} {detail}')
print(f'\nparquet  sha256: {PARQUET_SHA256}')
print(f'metadata sha256: {METADATA_SHA256}')
print(f'g = {G:.0f}, prior = {PRIOR_DESC}, models = {N_MODELS:,}, '
      f'load+validate {T_DATA_LOAD_S:.1f} s')
INPUT_VALIDATED = True
```

```python
# [CHECKPOINT TOOLS] Drive-free progress protection: package the current
# checkpoint state into a compact ZIP and download it to your computer;
# after a destroyed runtime, re-upload it with the next cell and resume.
# Bundles contain ONLY checkpoints + compatibility metadata — never the
# cloned repository, .git, dataset copies, caches, or credentials.
import hashlib, json, time, zipfile
import numpy as np

CKPT_BUNDLE_SCHEMA = 1

def _ckpt_stage_summary():
    out = {}
    for stage, path in (('PASS1', PASS1_CKPT), ('PASS2', PASS2_CKPT)):
        if path.exists():
            with np.load(path, allow_pickle=False) as z:
                out[stage] = {
                    'models_done': int(z['done']) if 'done' in z.files
                                   else int(z['models_done']),
                    'next_k': int(z['next_k']),
                    'next_rank': int(z['next_rank']),
                    'digest': str(z['digest']),
                }
    return out

def export_checkpoint_bundle():
    """Create WORK_DIR/checkpoint_bundle.zip from the current checkpoints."""
    stages = _ckpt_stage_summary()
    if not stages:
        print('no checkpoint files exist yet — nothing to export')
        return None
    manifest = dict(
        schema_version=CKPT_BUNDLE_SCHEMA, created_utc=time.strftime(
            '%Y-%m-%dT%H:%M:%SZ', time.gmtime()),
        parquet_sha256=PARQUET_SHA256, metadata_sha256=METADATA_SHA256,
        code_commit=RESOLVED_COMMIT, p=P, n=n, g=G,
        model_prior=list(MODEL_PRIOR), dtype=DTYPE,
        top_k=TOP_K_MODELS, grid_points=GRID_POINTS,
        compute_coef_densities=bool(COMPUTE_COEF_DENSITIES),
        stages=stages,
        note='resume compatibility is additionally enforced by the SHA-256 '
             'digest stored inside each .ckpt.npz')
    mpath = TMP_DIR / 'checkpoint_bundle_manifest.json'
    mpath.write_text(json.dumps(manifest, indent=2))
    zpath = WORK_DIR / 'checkpoint_bundle.zip'
    with zipfile.ZipFile(zpath, 'w', zipfile.ZIP_DEFLATED) as z:
        z.write(mpath, arcname='checkpoint_bundle_manifest.json')
        for path in (PASS1_CKPT, PASS2_CKPT):
            if path.exists():
                z.write(path, arcname=path.name)
        sidecar = CKPT_DIR / 'pass1_sidecar.json'
        if sidecar.exists():
            z.write(sidecar, arcname=sidecar.name)
    sha = hashlib.sha256(zpath.read_bytes()).hexdigest()
    print(f'bundle: {zpath} ({zpath.stat().st_size:,} B)\nsha256: {sha}')
    for stage, info in stages.items():
        print(f'  {stage}: {info["models_done"]:,} models done, '
              f'next k={info["next_k"]}, rank={info["next_rank"]:,}')
    return zpath

def download_checkpoint_bundle():
    """export_checkpoint_bundle() + browser download via Colab."""
    zpath = export_checkpoint_bundle()
    if zpath is None:
        return
    try:
        from google.colab import files
        files.download(str(zpath))
    except Exception as exc:
        print('browser download unavailable here:', exc)

print('checkpoint tools ready: call download_checkpoint_bundle() at any '
      'time to save progress to your computer (no Drive needed)')
```

```python
# [CHECKPOINT RESTORE — optional] Upload a previously downloaded
# checkpoint_bundle.zip and resume without any Google Drive account.
# Set the flag to True and run this cell; it validates the bundle and
# rejects anything incompatible instead of risking double-counting.
import hashlib, json, pathlib, zipfile
import numpy as np

RESTORE_CHECKPOINT_UPLOAD = False   # set True, run, and pick the .zip file

def _safe_extract_zip(zpath, dest):
    """Extract, refusing absolute paths and path traversal."""
    dest = dest.resolve()
    with zipfile.ZipFile(zpath) as z:
        for zi in z.infolist():
            name = zi.filename.replace('\\', '/')
            if name.startswith('/') or '..' in name.split('/'):
                raise ValueError(f'unsafe path in bundle: {zi.filename!r}')
            target = (dest / name).resolve()
            if not str(target).startswith(str(dest)):
                raise ValueError(f'path escapes destination: {zi.filename!r}')
        z.extractall(dest)

def restore_checkpoint_bundle(zpath):
    stage_dir = TMP_DIR / 'restore_upload'     # outside the git clone
    stage_dir.mkdir(parents=True, exist_ok=True)
    _safe_extract_zip(zpath, stage_dir)
    mpath = stage_dir / 'checkpoint_bundle_manifest.json'
    if not mpath.exists():
        raise ValueError('bundle has no checkpoint_bundle_manifest.json')
    man = json.loads(mpath.read_text())
    problems = []
    def need(field, expected):
        if man.get(field) != expected:
            problems.append(f'{field}: bundle {man.get(field)!r} '
                            f'!= run {expected!r}')
    need('schema_version', CKPT_BUNDLE_SCHEMA)
    need('parquet_sha256', PARQUET_SHA256)
    need('code_commit', RESOLVED_COMMIT)
    need('p', P); need('n', n); need('g', G)
    need('model_prior', list(MODEL_PRIOR)); need('dtype', DTYPE)
    need('top_k', TOP_K_MODELS); need('grid_points', GRID_POINTS)
    need('compute_coef_densities', bool(COMPUTE_COEF_DENSITIES))
    if problems:
        raise ValueError('INCOMPATIBLE checkpoint bundle — refusing to '
                         'restore:\n  ' + '\n  '.join(problems))
    restored = []
    for name in (PASS1_CKPT.name, PASS2_CKPT.name, 'pass1_sidecar.json'):
        src = stage_dir / name
        if src.exists():
            (CKPT_DIR / name).write_bytes(src.read_bytes())
            restored.append(name)
    print('restored:', ', '.join(restored))
    for stage, info in man['stages'].items():
        print(f'  {stage}: resumes after {info["models_done"]:,} models '
              f'(next k={info["next_k"]}, rank={info["next_rank"]:,}) — '
              'ranges before that point are never re-counted')
    print('resume: leave AUTO_RESUME = True and run the PASS cells; each '
          'checkpoint is re-verified against the data/prior SHA-256 digest '
          'before any model is scored')

if RESTORE_CHECKPOINT_UPLOAD:
    from google.colab import files
    up = files.upload()
    for fname in up:
        restore_checkpoint_bundle(pathlib.Path(fname))
else:
    print('restore idle (set RESTORE_CHECKPOINT_UPLOAD = True to upload a '
          'bundle after a destroyed runtime)')
```

```python
# [SMOKE TEST — MANDATORY] Reduced deterministic CPU/GPU parity on the
# canonical rows (p = SMOKE_P, validated in the local phase). The expensive
# cell refuses to run unless SMOKE_TEST_PASSED is True.
import numpy as np, time
from scipy.special import logsumexp
from gpubma.cpu.enumeration import enumerate_models
from gpubma.gpu.enumerator import enumerate_models_gpu
from gpubma.priors.model_priors import log_model_prior_function

assert INPUT_VALIDATED
t0 = time.perf_counter()
p_s = SMOKE_P
cols = [f'x{j}' for j in range(1, p_s + 1)]
Xs = DF[cols].to_numpy(np.float64)
Xs_r = Xs - Q @ (Q.T @ Xs)
lp_s, _ = log_model_prior_function(MODEL_PRIOR, p_s)
kw = dict(df_resid=n - 1, g=G, log_model_prior=lp_s,
          tss_norm=CONV['tss_norm'], k_always=2, compute_coefficients=True)
cpu = enumerate_models(Xs_r, Y_R, top_k=10, **kw)
gpu = enumerate_models_gpu(Xs_r, Y_R, top_k=10, keep_scores=True, **kw)

ATOL_SCORE, ATOL_PROB = 1e-9, 1e-10   # validated local-phase tolerances
rng = np.random.default_rng(SMOKE_RNG_SEED)
special = {'null': 0, 'full': (1 << p_s) - 1}
for j in range(p_s):
    special[f'singleton_{j}'] = 1 << j
k_c = p_s // 2
special['central_first'] = (1 << k_c) - 1
special['central_last'] = ((1 << k_c) - 1) << (p_s - k_c)
for i, m in enumerate(rng.integers(0, 1 << p_s, SMOKE_N_RANDOM_MODELS)):
    special[f'random_{i}'] = int(m)

gpu_pmp = np.exp(gpu['log_scores'] - logsumexp(gpu['log_scores']))
SMOKE = {
    'model_count': (cpu['n_models_evaluated'] == 2**p_s
                    and gpu['n_models_evaluated'] == 2**p_s),
    'all_scores': float(np.abs(cpu['log_scores'] - gpu['log_scores']).max()),
    'special_scores': max(abs(float(cpu['log_scores'][m] - gpu['log_scores'][m]))
                          for m in special.values()),
    'log_normalizer': abs(cpu['log_normalizer'] - gpu['log_normalizer']),
    'pmp': float(np.abs(cpu['pmp'] - gpu_pmp).max()),
    'pip': float(np.abs(cpu['pip'] - gpu['pip']).max()),
    'size_dist': float(np.abs(cpu['size_distribution']
                              - gpu['size_distribution']).max()),
    'coef_mean': float(np.abs(cpu['coef_mean'] - gpu['coef_mean']).max()),
    'coef_sd': float(np.abs(cpu['coef_sd'] - gpu['coef_sd']).max()),
    'top10_masks_equal': ([m['mask'] for m in cpu['top_models']]
                          == [m['mask'] for m in gpu['top_models']]),
    'finite': bool(np.isfinite(gpu['log_scores']).all()),
}
SMOKE_TEST_PASSED = bool(
    SMOKE['model_count'] and SMOKE['top10_masks_equal'] and SMOKE['finite']
    and SMOKE['all_scores'] < ATOL_SCORE
    and SMOKE['special_scores'] < ATOL_SCORE
    and SMOKE['log_normalizer'] < ATOL_SCORE
    and SMOKE['pmp'] < ATOL_PROB and SMOKE['pip'] < ATOL_PROB
    and SMOKE['size_dist'] < ATOL_PROB
    and SMOKE['coef_mean'] < ATOL_SCORE and SMOKE['coef_sd'] < ATOL_SCORE)
T_SMOKE_S = time.perf_counter() - t0
print(f'smoke test p={p_s}: {2**p_s:,} models, {len(special)} named IDs, '
      f'{T_SMOKE_S:.1f} s')
for k, v in SMOKE.items():
    print(f'  {k:16s}: {v if isinstance(v, bool) else f"{v:.3e}"}')
print('SMOKE_TEST_PASSED =', SMOKE_TEST_PASSED)
if not SMOKE_TEST_PASSED:
    raise AssertionError('CPU/GPU parity smoke test FAILED — do not run the '
                         'full enumeration; investigate first.')
```

```python
# [PASS2] Bounded-memory exact second sweep over ALL 2^30 models (weights =
# exact global PMPs from the saved PASS1 normalizer). Produces what PASS1's
# O(p) accumulators cannot: the top-K model table (K = TOP_K_MODELS), the
# per-family joint posterior (true/proxy both/only/neither), the EXACT
# posterior rank of the true model, score extrema / non-finite counts,
# coefficient sign probabilities, and (optionally) exact conditional
# coefficient density grids. Deterministic, checkpointed, float64.
#
# Sign probabilities use the Gaussian limit of the t_(n-1) conditional
# posterior (n - 1 = 1999): absolute error < 1e-3, labelled 'approx' in the
# output schema. Everything else in this pass is exact.
import hashlib, json, math, time
import numpy as np
import torch

if PASS1_RESULT is None or not RUN_PASS2:
    print('PASS2 skipped (needs PASS1 results and RUN_PASS2=True)')
    PASS2_DONE = False
else:
    from gpubma.gpu.enumerator import (binomial_table, unrank_combinations,
                                       _save_checkpoint, _load_checkpoint)
    dev = torch.device('cuda')
    LOG_NORM = float(PASS1_RESULT['log_normalizer'])
    PIP1 = np.asarray(PASS1_RESULT['pip'])
    CM1 = np.asarray(PASS1_RESULT['coef_mean'])
    CS1 = np.asarray(PASS1_RESULT['coef_sd'])
    Zxx_np, Zxy_np = X_r.T @ X_r, X_r.T @ Y_R
    tss = float(Y_R @ Y_R)
    tss_norm, k_alw, dfree = CONV['tss_norm'], CONV['k_always'], float(n - 1)
    ess_always = tss_norm - tss
    lp_np = np.array([LOG_PRIOR(k) for k in range(P + 1)])
    s_g = G / (1.0 + G)
    log1pg = math.log1p(G)
    tiny = float(np.finfo(np.float64).tiny)
    c_t = (math.lgamma((dfree + 1) / 2) - math.lgamma(dfree / 2)
           - 0.5 * math.log(dfree * math.pi))
    PASS2_CHUNK = 8192

    # exact log score of the TRUE model (x1..x15) computed up front
    from scipy.linalg import cho_factor, cho_solve
    idx_t = np.arange(15)
    cft = cho_factor(Zxx_np[np.ix_(idx_t, idx_t)], lower=True)
    ess_t = float(Zxy_np[idx_t] @ cho_solve(cft, Zxy_np[idx_t]))
    omr_t = max((tss - ess_t) / tss_norm, tiny)
    TRUE_SCORE = (0.5 * (dfree - 15 - k_alw) * log1pg
                  - 0.5 * dfree * math.log1p(G * omr_t) + lp_np[15])

    def grid_np():
        pip_safe = np.maximum(PIP1, 1e-300)
        cm = CM1 / pip_safe
        cv = np.maximum((CS1**2 + CM1**2) / pip_safe - cm**2, 1e-30)
        cs = np.sqrt(cv)
        return np.linspace(cm - 8 * cs, cm + 8 * cs, GRID_POINTS, axis=1)

    GRID = grid_np()
    h = hashlib.sha256()
    for arr in (Zxx_np, Zxy_np, lp_np, GRID):
        h.update(np.ascontiguousarray(arr, np.float64).tobytes())
    h.update(np.array([tss, tss_norm, dfree, G, k_alw, LOG_NORM, TRUE_SCORE,
                       TOP_K_MODELS, GRID_POINTS,
                       float(COMPUTE_COEF_DENSITIES)], np.float64).tobytes())
    digest = h.hexdigest()

    Zxx = torch.from_numpy(Zxx_np).to(dev)
    Zxy = torch.from_numpy(Zxy_np).to(dev)
    grid_t = torch.from_numpy(GRID).to(dev)
    binom_np = binomial_table(P)
    binom = torch.from_numpy(binom_np).to(dev)
    jbits = torch.arange(P, dtype=torch.int64, device=dev)

    st0 = dict(
        dens=torch.zeros(P, GRID_POINTS, dtype=torch.float64, device=dev),
        top_s=torch.full((TOP_K_MODELS,), -math.inf, dtype=torch.float64,
                         device=dev),
        top_m=torch.zeros(TOP_K_MODELS, dtype=torch.int64, device=dev),
        sum_w=torch.zeros((), dtype=torch.float64, device=dev),
        fam=torch.zeros(15, 3, dtype=torch.float64, device=dev),  # both/T-only/P-only
        sign_num=torch.zeros(P, dtype=torch.float64, device=dev),
        n_higher=torch.zeros((), dtype=torch.float64, device=dev),
        n_tied=torch.zeros((), dtype=torch.float64, device=dev),
        mass_ge=torch.zeros((), dtype=torch.float64, device=dev),
        smin=torch.tensor(math.inf, dtype=torch.float64, device=dev),
        smax=torch.tensor(-math.inf, dtype=torch.float64, device=dev),
        nonfinite=torch.zeros((), dtype=torch.float64, device=dev),
    )
    done, k0, r0 = 0, 0, 0
    if AUTO_RESUME and PASS2_CKPT.exists():
        cp = _load_checkpoint(PASS2_CKPT)
        assert str(cp['digest']) == digest, ('PASS2 checkpoint belongs to a '
                                             'different configuration; refusing')
        for key in st0:
            st0[key] = torch.from_numpy(np.atleast_1d(cp[key])).to(dev).reshape(
                st0[key].shape).clone()
        done, k0, r0 = int(cp['done']), int(cp['next_k']), int(cp['next_rank'])
        print(f'PASS2 resuming at k={k0}, rank={r0:,} ({done:,} models done)')

    def save_ckpt(nk, nr):
        state = {key: v.cpu().numpy() for key, v in st0.items()}
        state.update(digest=np.str_(digest), done=np.int64(done),
                     next_k=np.int64(nk), next_rank=np.int64(nr))
        _save_checkpoint(PASS2_CKPT, state)

    t0 = time.time()
    last_ck = last_pr = t0
    if PASS2_CKPT.exists() and done >= N_MODELS:
        print('PASS2 already complete in checkpoint')
    for k in range(k0, P + 1):
        total_k = int(binom_np[P, k])
        r = r0 if k == k0 else 0
        while r < total_k and done < N_MODELS:
            B = min(PASS2_CHUNK, total_k - r)
            if k == 0:
                omr = max(tss / tss_norm, tiny)
                sc = torch.tensor([0.5 * (dfree - k_alw) * log1pg
                                   - 0.5 * dfree * math.log1p(G * omr)
                                   + lp_np[0]], dtype=torch.float64, device=dev)
                masks = torch.zeros(1, dtype=torch.int64, device=dev)
                w = torch.exp(sc - LOG_NORM)
            else:
                ranks = torch.arange(r, r + B, dtype=torch.int64, device=dev)
                idx = unrank_combinations(ranks, k, binom, torch)
                Z = Zxx[idx.unsqueeze(2), idx.unsqueeze(1)]
                b = Zxy[idx].unsqueeze(-1)
                L = torch.linalg.cholesky(Z)
                u = torch.linalg.solve_triangular(L, b, upper=False)
                ess = (u.squeeze(-1) ** 2).sum(dim=1)
                omr = torch.clamp((tss - ess) / tss_norm, min=tiny)
                sc = (0.5 * (dfree - k - k_alw) * log1pg
                      - 0.5 * dfree * torch.log1p(G * omr) + lp_np[k])
                masks = (torch.ones_like(idx) << idx).sum(dim=1)
                w = torch.exp(sc - LOG_NORM)
                beta_hat = torch.linalg.solve_triangular(
                    L.transpose(1, 2), u, upper=True).squeeze(-1)
                zinv = torch.cholesky_inverse(L).diagonal(dim1=-2, dim2=-1)
                b_gam = tss_norm - s_g * (ess_always + ess)
                loc = s_g * beta_hat
                scale = torch.sqrt((b_gam / dfree).unsqueeze(1) * s_g * zinv)
                onehot = torch.nn.functional.one_hot(idx, P).to(torch.float64)
                # sign prob (normal limit of t_1999): Phi(loc/scale)
                phi = 0.5 * (1.0 + torch.erf(loc / scale / math.sqrt(2.0)))
                st0['sign_num'] += torch.einsum('bkp,bk->p', onehot,
                                                w.unsqueeze(1) * phi)
                if COMPUTE_COEF_DENSITIES:
                    xg = grid_t[idx]
                    zz = (xg - loc.unsqueeze(-1)) / scale.unsqueeze(-1)
                    logpdf = (c_t - torch.log(scale).unsqueeze(-1)
                              - 0.5 * (dfree + 1) * torch.log1p(zz * zz / dfree))
                    st0['dens'] += torch.einsum(
                        'bkp,bkg->pg', onehot,
                        w.unsqueeze(-1).unsqueeze(-1) * torch.exp(logpdf))
            bits = ((masks.unsqueeze(1) >> jbits) & 1).to(torch.float64)
            tb, pb = bits[:, :15], bits[:, 15:]
            st0['fam'][:, 0] += (w.unsqueeze(1) * tb * pb).sum(dim=0)
            st0['fam'][:, 1] += (w.unsqueeze(1) * tb * (1 - pb)).sum(dim=0)
            st0['fam'][:, 2] += (w.unsqueeze(1) * (1 - tb) * pb).sum(dim=0)
            st0['sum_w'] += w.sum()
            st0['n_higher'] += (sc > TRUE_SCORE).sum()
            st0['n_tied'] += (sc == TRUE_SCORE).sum()
            st0['mass_ge'] += (w * (sc >= TRUE_SCORE)).sum()
            finite = torch.isfinite(sc)
            st0['nonfinite'] += (~finite).sum()
            st0['smin'] = torch.minimum(st0['smin'], sc[finite].min())
            st0['smax'] = torch.maximum(st0['smax'], sc[finite].max())
            cs_ = torch.cat([st0['top_s'], sc])
            cm_ = torch.cat([st0['top_m'], masks])
            best = torch.topk(cs_, k=TOP_K_MODELS)
            st0['top_s'], st0['top_m'] = best.values, cm_[best.indices]
            done += B
            r += B
            now = time.time()
            if now - last_pr >= PROGRESS_EVERY_S:
                rate = done / max(now - t0, 1e-9)
                print(f'[PASS2 {done / N_MODELS:7.2%}] layer k={k:2d}  '
                      f'{done:,}/{N_MODELS:,}  {rate:,.0f}/s  '
                      f'ETA {(N_MODELS - done) / max(rate, 1e-9):,.0f}s',
                      flush=True)
                last_pr = now
            if now - last_ck >= CHECKPOINT_EVERY_S:
                save_ckpt(k if r < total_k else k + 1, r if r < total_k else 0)
                last_ck = now
    assert done == N_MODELS, f'PASS2 incomplete: {done:,}'
    total_w = float(st0['sum_w'].cpu())
    assert abs(total_w - 1.0) < 1e-9, f'sum of exact PMPs = {total_w!r}'
    save_ckpt(P + 1, 0)
    T_PASS2_S = time.time() - t0
    order = torch.argsort(st0['top_s'], descending=True)
    TOP_SCORES = st0['top_s'][order].cpu().numpy()
    TOP_MASKS = st0['top_m'][order].cpu().numpy()
    FAM = st0['fam'].cpu().numpy()          # (15, 3): both, true-only, proxy-only
    SIGN_NUM = st0['sign_num'].cpu().numpy()
    DENS = st0['dens'].cpu().numpy() if COMPUTE_COEF_DENSITIES else None
    TRUE_RANK = int(float(st0['n_higher'].cpu())) + 1
    TRUE_TIES = int(float(st0['n_tied'].cpu())) - 1
    MASS_GE_TRUE = float(st0['mass_ge'].cpu())
    SCORE_MIN = float(st0['smin'].cpu())
    SCORE_MAX = float(st0['smax'].cpu())
    N_NONFINITE = int(float(st0['nonfinite'].cpu()))
    PASS2_DONE = True
    print(f'PASS2 complete: {done:,} models, sum PMP = {total_w:.15f}, '
          f'{T_PASS2_S:,.0f} s this session; true-model exact rank = '
          f'{TRUE_RANK:,} (ties: {TRUE_TIES})')
```

```python
# [PLOT-PMP] Posterior model probabilities — bmagraph pmp equivalent.
def bma_pmp_plot(top_n=None, coverage=None, normalize_display=False):
    assert (top_n is None) != (coverage is None)

    if coverage is not None:
        K = int(np.searchsorted(
            TOPK.cumulative_pmp_global.to_numpy(),
            coverage
        ) + 1)

        if K > len(TOPK):
            raise ValueError(
                f'coverage {coverage} requires more than the '
                f'{len(TOPK)} retained models'
            )

        sel_label = (
            f'minimal K = {K} reaching coverage >= {coverage}'
        )
    else:
        K = min(int(top_n), len(TOPK))
        sel_label = f'top {K} models'

    t = TOPK.head(K).copy()

    other = float(np.clip(
        1.0 - t.model_pmp_global.sum(),
        0.0,
        1.0
    ))

    shown = np.append(
        t.model_pmp_global.to_numpy(),
        other
    )

    if abs(shown.sum() - 1.0) > 1e-9:
        raise AssertionError(
            'displayed models + Other must sum to one'
        )

    if normalize_display:
        displayed_mass = shown[:-1].sum()

        if displayed_mass <= 0:
            raise ValueError(
                'displayed model mass must be positive'
            )

        disp = shown / displayed_mass
    else:
        disp = shown

    fig_width = max(10.0, 0.055 * K + 3.0)

    fig, ax = plt.subplots(
        figsize=(fig_width, 5.6)
    )

    xs = np.arange(1, K + 1)

    ax.bar(
        xs,
        disp[:-1],
        width=0.85,
        color='#1f77b4',
        linewidth=0,
        label=(
            f'exact global PMP '
            f'(denominator: all $2^{{{P}}}$ models)'
        )
    )

    other_x = K + max(3.0, 0.025 * K)

    ax.bar(
        [other_x],
        [disp[-1]],
        width=max(1.0, 0.008 * K),
        color='0.6',
        hatch='//',
        edgecolor='0.25',
        linewidth=0.6,
        label=(
            'Other models '
            f'(exact omitted mass = {other:.6f})'
        )
    )

    ax2 = ax.twinx()

    ax2.plot(
        xs,
        t.cumulative_pmp_global,
        color='#d62728',
        marker='.',
        markersize=2.3,
        linewidth=1,
        label='cumulative global PMP'
    )

    ax2.set_ylabel('cumulative global PMP')
    ax2.set_ylim(0, 1.02)
    ax2.grid(False)

    ax.set_xlim(
        0,
        other_x + max(2.0, 0.015 * K)
    )

    ax.set_xlabel(
        'model rank (descending global PMP)',
        labelpad=8
    )

    ax.set_ylabel(
        'displayed share (renormalized — NOT probabilities)'
        if normalize_display
        else 'posterior model probability'
    )

    title = (
        f'Posterior model probabilities, '
        f'p = {P} ({sel_label})'
    )

    if normalize_display:
        title += (
            ' [display-normalized mode — '
            'bars are not global PMPs]'
        )

    ax.set_title(title, pad=10)

    # Combined graphical legend below the x-axis.
    h1, l1 = ax.get_legend_handles_labels()
    h2, l2 = ax2.get_legend_handles_labels()

    fig.legend(
        h1 + h2,
        l1 + l2,
        loc='lower center',
        bbox_to_anchor=(0.5, 0.015),
        ncol=3,
        fontsize=8,
        frameon=False,
        columnspacing=1.6,
        handletextpad=0.7,
        borderaxespad=0.0
    )

    fig.subplots_adjust(
        left=0.075,
        right=0.925,
        top=0.90,
        bottom=0.20
    )

    t['included_variables'] = t.model_mask.map(
        lambda m: ' '.join(mask_vars(m))
    )

    t['other_model_mass'] = other

    # Text prepared for notebook output, not embedded in the image.
    top5_lines = [
        (
            f'#{int(r.model_rank)}: '
            f'{{{", ".join(mask_vars(r.model_mask))}}}  '
            f'PMP={r.model_pmp_global:.4f}  '
            f'cum={r.cumulative_pmp_global:.4f}'
        )
        for r in t.head(5).itertuples()
    ]

    top5_text = '\n'.join(top5_lines)

    return fig, t, top5_text


# Canonical PMP figure: top 200 models
fig, tbl, top5_text = bma_pmp_plot(top_n=200)

save_fig(fig, 'pmp_top200')
save_table(tbl, 'pmp_top200')

plt.show()

# Printed separately in the notebook; not included in the saved figure.
print('\nTop 5 posterior models')
print('----------------------')
print(top5_text)
```

```python
# [PLOT-VARMAP] Coverage-based variable-inclusion map — bmagraph varmap equivalent.
from scipy.linalg import cho_factor, cho_solve
from matplotlib.patches import Patch

PMP_COVERAGE_TARGET = 0.90
COEF_ZERO_TOL = 1e-12

# ------------------------------------------------------------------
# 1. Select the minimal number of models reaching the target coverage
# ------------------------------------------------------------------
cum = TOPK.cumulative_pmp_global.to_numpy(dtype=np.float64)

K_COV = int(
    np.searchsorted(
        cum,
        PMP_COVERAGE_TARGET,
        side='left'
    ) + 1
)

if K_COV > len(TOPK):
    raise ValueError(
        f'target {PMP_COVERAGE_TARGET} requires more than '
        f'{len(TOPK)} retained models'
    )

sel = TOPK.head(K_COV).copy()

CPMP_K = float(sel.model_pmp_global.sum())
OTHER_MASS = float(
    np.clip(1.0 - CPMP_K, 0.0, 1.0)
)

print(
    f'coverage target {PMP_COVERAGE_TARGET}: '
    f'minimal K = {K_COV}, '
    f'achieved CPMP = {CPMP_K:.6f}, '
    f'omitted models = {N_MODELS - K_COV:,}, '
    f'exact omitted mass = {OTHER_MASS:.6f}'
)

# ------------------------------------------------------------------
# 2. Recover the exact regression inputs used in the GPUBMA run
# ------------------------------------------------------------------
if 'X_r' not in globals():
    raise NameError(
        '`X_r` is unavailable. Re-run the preprocessing cell.'
    )

X_varmap = np.asarray(
    X_r,
    dtype=np.float64
)

if X_varmap.ndim != 2:
    raise ValueError(
        f'X_r must be two-dimensional; got {X_varmap.shape}'
    )

if X_varmap.shape[1] != P:
    raise ValueError(
        f'X_r has {X_varmap.shape[1]} columns but P = {P}'
    )

n_obs = X_varmap.shape[0]
y_varmap = None
y_source = None

for candidate_name in (
    'y_r',
    'yc',
    'y_centered',
    'y_c',
    'y_np',
    'Y_r',
    'Y_centered',
    'Y',
    'y'
):
    if candidate_name not in globals():
        continue

    candidate = np.asarray(
        globals()[candidate_name],
        dtype=np.float64
    ).reshape(-1)

    if candidate.size == n_obs:
        y_varmap = candidate
        y_source = candidate_name
        break

if y_varmap is None:
    raise NameError(
        f'Could not locate a response vector with {n_obs} observations.'
    )

if not np.isfinite(X_varmap).all():
    raise ValueError('X_r contains non-finite values')

if not np.isfinite(y_varmap).all():
    raise ValueError(
        f'{y_source} contains non-finite values'
    )

print(
    f'varmap regression inputs: '
    f'X source = X_r, shape = {X_varmap.shape}; '
    f'y source = {y_source}, shape = {y_varmap.shape}'
)

# ------------------------------------------------------------------
# 3. Conditional posterior means under the fixed g-prior
# ------------------------------------------------------------------
Zxx_np = X_varmap.T @ X_varmap
Zxy_np = X_varmap.T @ y_varmap

S_G = G / (1.0 + G)

cond_coef = np.zeros(
    (K_COV, P),
    dtype=np.float64
)

incl = np.zeros(
    (K_COV, P),
    dtype=bool
)

for i, model_mask in enumerate(sel.model_mask):
    jj = np.array(
        [
            j for j in range(P)
            if (int(model_mask) >> j) & 1
        ],
        dtype=int
    )

    if jj.size == 0:
        continue

    incl[i, jj] = True

    gram_sub = Zxx_np[np.ix_(jj, jj)]
    cross_sub = Zxy_np[jj]

    chol, lower = cho_factor(
        gram_sub,
        lower=True,
        check_finite=False
    )

    beta_ols = cho_solve(
        (chol, lower),
        cross_sub,
        check_finite=False
    )

    cond_coef[i, jj] = S_G * beta_ols

print('Conditional coefficients calculated.')

# Variables ordered from highest to lowest exact PIP
row_order = np.argsort(-np.asarray(PIP))


# ------------------------------------------------------------------
# 4. Stata-style variable-inclusion map
# ------------------------------------------------------------------
def varmap_figure(
    proportional=True,
    max_cols=None
):
    cols = (
        min(K_COV, int(max_cols))
        if max_cols is not None
        else K_COV
    )

    model_pmp = sel.model_pmp_global.to_numpy(
        dtype=np.float64
    )[:cols]

    if proportional:
        widths = model_pmp
    else:
        displayed_mass = float(model_pmp.sum())
        widths = np.full(
            cols,
            displayed_mass / cols,
            dtype=np.float64
        )

    xedges = np.concatenate(
        ([0.0], np.cumsum(widths))
    )

    displayed_mass = float(widths.sum())
    omitted_mass_display = float(
        np.clip(1.0 - displayed_mass, 0.0, 1.0)
    )

    # Layout close to the canonical bmagraph varmap style
    fig, (ax, axp) = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(12.8, 11.8),
        sharey=True,
        gridspec_kw={
            'width_ratios': [5.6, 1.1],
            'wspace': 0.20
        }
    )

    positive_color = '#2166ac'
    negative_color = '#b2182b'
    near_zero_color = '#ffee99'
    omitted_color = '0.86'

    # Horizontal row guides
    for ri in range(P + 1):
        ax.axhline(
            ri,
            color='0.85',
            linewidth=0.45,
            zorder=0
        )
        axp.axhline(
            ri,
            color='0.88',
            linewidth=0.45,
            zorder=0
        )

    # Inclusion cells
    for ri, j in enumerate(row_order):
        for ci in range(cols):
            if not incl[ci, j]:
                continue

            value = cond_coef[ci, j]

            if abs(value) < COEF_ZERO_TOL:
                facecolor = near_zero_color
            elif value > 0:
                facecolor = positive_color
            else:
                facecolor = negative_color

            ax.add_patch(
                plt.Rectangle(
                    (xedges[ci], ri),
                    widths[ci],
                    1.0,
                    facecolor=facecolor,
                    edgecolor='none',
                    zorder=2
                )
            )

    # Fine white borders separating displayed models
    for edge in xedges[1:-1]:
        ax.axvline(
            edge,
            color='white',
            linewidth=0.35,
            alpha=0.9,
            zorder=3
        )

    # Exact omitted posterior mass
    other_left = xedges[-1]

    ax.add_patch(
        plt.Rectangle(
            (other_left, 0),
            omitted_mass_display,
            P,
            facecolor=omitted_color,
            edgecolor='0.45',
            linewidth=0.8,
            hatch='///',
            zorder=1
        )
    )

    if cols == K_COV:
        omitted_label = (
            'Other models\n'
            f'exact mass = {OTHER_MASS:.4f}\n'
            f'{N_MODELS - K_COV:,} models omitted'
        )
    else:
        omitted_label = (
            'Models not shown\n'
            f'total mass = {omitted_mass_display:.4f}\n'
            f'{N_MODELS - cols:,} models'
        )

    if omitted_mass_display > 0.025:
        ax.text(
            other_left + omitted_mass_display / 2,
            P / 2,
            omitted_label,
            rotation=90,
            ha='center',
            va='center',
            fontsize=8,
            zorder=4
        )

    # Main map axes
    ax.set_xlim(0.0, 1.0)
    ax.set_ylim(P, 0)

    ax.set_yticks(
        np.arange(P) + 0.5
    )

    ax.set_yticklabels(
        [predictors[j] for j in row_order],
        fontsize=8
    )

    ax.set_xlabel(
        'cumulative global PMP '
        '(column width proportional to exact global PMP)'
        if proportional
        else
        'displayed models at equal width '
        '(column widths are not proportional to PMP)'
    )

    truncation_text = (
        f', first {cols} of K = {K_COV} models shown'
        if cols < K_COV
        else ''
    )

    ax.set_title(
        f'Variable-inclusion map, p = {P}: '
        f'minimal K = {K_COV} models with cumulative global PMP '
        f'{CPMP_K:.4f} >= target {PMP_COVERAGE_TARGET}'
        f'{truncation_text}\n'
        'blue = included, positive conditional coefficient; '
        'red = included, negative; '
        f'pale yellow = |coef| < {COEF_ZERO_TOL:.0e}; '
        'white = excluded',
        fontsize=13,
        pad=10
    )

    # Exact PIP panel
    y_positions = np.arange(P) + 0.5
    ordered_pip = np.asarray(PIP)[row_order]

    axp.barh(
        y_positions,
        ordered_pip,
        color='#1f77b4',
        height=0.78
    )

    axp.set_xlim(0.0, 1.10)
    axp.set_xlabel('PIP (exact)')
    axp.tick_params(
        axis='y',
        labelleft=False
    )

    for ri, pip_value in enumerate(ordered_pip):
        axp.text(
            min(float(pip_value) + 0.012, 1.065),
            ri + 0.5,
            f'{pip_value:.2f}',
            va='center',
            ha='left',
            fontsize=7,
            clip_on=False
        )

    # Keep the shared rows perfectly aligned
    axp.set_ylim(P, 0)

    fig.subplots_adjust(
        left=0.18,
        right=0.96,
        top=0.92,
        bottom=0.08
    )

    return fig


# ------------------------------------------------------------------
# 5. Canonical proportional-width figure
# ------------------------------------------------------------------
fig_full = varmap_figure(
    proportional=True
)

plt.show()
save_fig(
    fig_full,
    'varmap_proportional'
)
plt.close(fig_full)


# ------------------------------------------------------------------
# 6. Equal-width diagnostic version
# ------------------------------------------------------------------
fig_equal = varmap_figure(
    proportional=False
)

save_fig(
    fig_equal,
    'varmap_equal_width'
)
plt.close(fig_equal)


# ------------------------------------------------------------------
# 7. Optional compact version when many columns are required
# ------------------------------------------------------------------
if K_COV > 60:
    fig_compact = varmap_figure(
        proportional=True,
        max_cols=60
    )

    save_fig(
        fig_compact,
        'varmap_compact_truncated'
    )
    plt.close(fig_compact)


# ------------------------------------------------------------------
# 8. Export the underlying long-format table
# ------------------------------------------------------------------
rows = []

for ci, model_row in enumerate(sel.itertuples()):
    for j in range(P):
        value = cond_coef[ci, j]
        is_included = bool(incl[ci, j])

        if not is_included or abs(value) < COEF_ZERO_TOL:
            coefficient_sign = 0
        elif value > 0:
            coefficient_sign = 1
        else:
            coefficient_sign = -1

        rows.append({
            'predictor': predictors[j],
            'predictor_index': int(j),
            'pip': float(PIP[j]),
            'model_rank': int(model_row.model_rank),
            'model_mask': int(model_row.model_mask),
            'model_size': int(model_row.model_size),
            'model_pmp_global': float(
                model_row.model_pmp_global
            ),
            'cumulative_pmp_global': float(
                model_row.cumulative_pmp_global
            ),
            'included': is_included,
            'conditional_coefficient': (
                float(value)
                if is_included
                else np.nan
            ),
            'coefficient_sign': coefficient_sign,
            'coverage_target': PMP_COVERAGE_TARGET,
            'achieved_coverage': CPMP_K,
            'other_model_mass': OTHER_MASS
        })

varmap_table = pd.DataFrame(rows)

save_table(
    varmap_table,
    f'varmap_coverage{int(PMP_COVERAGE_TARGET * 100)}'
)
```

```python
# [PLOT-COEFSUMMARY] Ordered exact BMA coefficient summary:
# left panel  = unconditional posterior mean with exact 95% equal-tail
#               credible interval from the full marginal BMA posterior
# right panel = exact global PIP
#
# The marginal posterior includes:
#   (1 - PIP) point mass at zero
#   + PIP × conditional-on-inclusion Student-t mixture
#
# Shared ordering:
#   1. predictors with PIP >= prior inclusion probability
#   2. predictors with PIP < prior inclusion probability
# Within each block:
#   largest unconditional posterior mean to smallest

from matplotlib.lines import Line2D
from scipy.integrate import cumulative_trapezoid

PIP_HIGH_THRESHOLD = 0.50
SIGN_STABILITY_THRESHOLD = 0.95

CREDIBLE_LEVEL = 0.95
ALPHA = 1.0 - CREDIBLE_LEVEL
LOWER_PROB = ALPHA / 2.0
UPPER_PROB = 1.0 - ALPHA / 2.0

# ------------------------------------------------------------------
# 1. Recover conditional sign probabilities in predictor order
# ------------------------------------------------------------------
SIGN_PROB_COLUMN = 'sign_prob_positive_given_incl_approx'

if SIGN_PROB_COLUMN not in pip_df.columns:
    raise KeyError(
        f'Column {SIGN_PROB_COLUMN!r} was not found in pip_df. '
        f'Available columns: {list(pip_df.columns)}'
    )

sign_prob_pos = (
    pip_df
    .set_index('variable')
    .loc[predictors, SIGN_PROB_COLUMN]
    .to_numpy(dtype=np.float64)
)

if sign_prob_pos.shape != (P,):
    raise ValueError(
        f'sign probability vector has shape {sign_prob_pos.shape}; '
        f'expected ({P},)'
    )

# ------------------------------------------------------------------
# 2. Recover exact conditional coefficient densities from PASS2
#
# Supports both notebook layouts:
#   A) DENS is a dictionary with keys 'grid' and 'conditional_density'
#   B) GRID and DENS are already numerical arrays with shape (P, G)
# ------------------------------------------------------------------
if 'DENS' not in globals():
    raise NameError(
        '`DENS` is unavailable. Run PASS2 or reload the '
        'coefficient-density results.'
    )

if isinstance(DENS, dict):
    if 'grid' not in DENS:
        raise KeyError(
            "`DENS['grid']` is unavailable."
        )

    if 'conditional_density' not in DENS:
        raise KeyError(
            "`DENS['conditional_density']` is unavailable. "
            f'Available keys: {list(DENS.keys())}'
        )

    coef_grid = np.asarray(
        DENS['grid'],
        dtype=np.float64
    )

    coef_cond_density = np.asarray(
        DENS['conditional_density'],
        dtype=np.float64
    )

else:
    if 'GRID' not in globals():
        raise NameError(
            '`DENS` is an array, so a matching `GRID` array is required.'
        )

    coef_grid = np.asarray(
        GRID,
        dtype=np.float64
    )

    coef_cond_density = np.asarray(
        DENS,
        dtype=np.float64
    )

print(
    'Coefficient-density arrays:',
    f'grid shape = {coef_grid.shape};',
    f'conditional density shape = {coef_cond_density.shape}'
)

if coef_grid.ndim != 2:
    raise ValueError(
        f'coefficient grid must be two-dimensional; '
        f'received shape {coef_grid.shape}'
    )

if coef_cond_density.ndim != 2:
    raise ValueError(
        f'conditional density must be two-dimensional; '
        f'received shape {coef_cond_density.shape}'
    )

if coef_grid.shape != coef_cond_density.shape:
    raise ValueError(
        f'coefficient grid shape {coef_grid.shape} does not match '
        f'conditional-density shape {coef_cond_density.shape}'
    )

if coef_grid.shape[0] != P:
    raise ValueError(
        f'density arrays contain {coef_grid.shape[0]} variables, '
        f'but P = {P}'
    )

if not np.isfinite(coef_grid).all():
    raise ValueError(
        'coefficient grid contains non-finite values'
    )

if not np.isfinite(coef_cond_density).all():
    raise ValueError(
        'conditional coefficient densities contain non-finite values'
    )

# ------------------------------------------------------------------
# 3. Exact marginal posterior quantiles, including point mass at zero
# ------------------------------------------------------------------
def conditional_quantile(grid, cdf, probability):
    probability = float(np.clip(probability, 0.0, 1.0))

    if probability <= 0.0:
        return float(grid[0])

    if probability >= 1.0:
        return float(grid[-1])

    return float(
        np.interp(
            probability,
            cdf,
            grid
        )
    )


def exact_marginal_quantile(
    grid,
    conditional_density,
    pip,
    probability
):
    """
    Quantile of:
        (1 - pip) * delta_0
        + pip * conditional continuous posterior.
    """
    grid = np.asarray(
        grid,
        dtype=np.float64
    )

    density = np.asarray(
        conditional_density,
        dtype=np.float64
    )

    if grid.ndim != 1 or density.ndim != 1:
        raise ValueError(
            'grid and density must be one-dimensional'
        )

    if grid.size != density.size:
        raise ValueError(
            'grid and density lengths do not match'
        )

    if np.any(np.diff(grid) <= 0):
        raise ValueError(
            'coefficient grid must be strictly increasing'
        )

    pip = float(np.clip(pip, 0.0, 1.0))
    probability = float(np.clip(probability, 0.0, 1.0))

    if pip <= 0.0:
        return 0.0

    density = np.clip(
        density,
        0.0,
        np.inf
    )

    area = float(
        np.trapezoid(
            density,
            grid
        )
    )

    if not np.isfinite(area) or area <= 0.0:
        raise ValueError(
            'conditional coefficient density has non-positive mass'
        )

    density = density / area

    conditional_cdf = cumulative_trapezoid(
        density,
        grid,
        initial=0.0
    )

    conditional_cdf = np.maximum.accumulate(
        conditional_cdf
    )

    if conditional_cdf[-1] <= 0.0:
        raise ValueError(
            'conditional CDF has zero terminal mass'
        )

    conditional_cdf /= conditional_cdf[-1]
    conditional_cdf[-1] = 1.0

    conditional_cdf_zero = float(
        np.interp(
            0.0,
            grid,
            conditional_cdf,
            left=0.0,
            right=1.0
        )
    )

    cdf_before_zero = pip * conditional_cdf_zero
    cdf_after_zero = cdf_before_zero + (1.0 - pip)

    if probability < cdf_before_zero:
        conditional_target = probability / pip

        return conditional_quantile(
            grid,
            conditional_cdf,
            conditional_target
        )

    if probability <= cdf_after_zero:
        return 0.0

    conditional_target = (
        probability - (1.0 - pip)
    ) / pip

    return conditional_quantile(
        grid,
        conditional_cdf,
        conditional_target
    )


COEF_Q025 = np.empty(
    P,
    dtype=np.float64
)

COEF_Q500 = np.empty(
    P,
    dtype=np.float64
)

COEF_Q975 = np.empty(
    P,
    dtype=np.float64
)

for j in range(P):
    COEF_Q025[j] = exact_marginal_quantile(
        coef_grid[j],
        coef_cond_density[j],
        PIP[j],
        LOWER_PROB
    )

    COEF_Q500[j] = exact_marginal_quantile(
        coef_grid[j],
        coef_cond_density[j],
        PIP[j],
        0.50
    )

    COEF_Q975[j] = exact_marginal_quantile(
        coef_grid[j],
        coef_cond_density[j],
        PIP[j],
        UPPER_PROB
    )

if np.any(COEF_Q025 > COEF_Q975):
    raise AssertionError(
        'at least one lower posterior quantile exceeds its upper quantile'
    )

INTERVAL_EXCLUDES_ZERO = (
    (COEF_Q025 > 0.0) |
    (COEF_Q975 < 0.0)
)

# ------------------------------------------------------------------
# 4. Shared display order
# ------------------------------------------------------------------
high_idx = np.where(
    PIP >= PIP_HIGH_THRESHOLD
)[0]

low_idx = np.where(
    PIP < PIP_HIGH_THRESHOLD
)[0]

high_order = high_idx[
    np.argsort(-COEF_MEAN[high_idx])
]

low_order = low_idx[
    np.argsort(-COEF_MEAN[low_idx])
]

order = np.concatenate([
    high_order,
    low_order
])

n_high = len(high_order)
y_positions = np.arange(P)

# ------------------------------------------------------------------
# 5. Sign classification
# ------------------------------------------------------------------
def sign_class(j):
    if PIP[j] < PIP_HIGH_THRESHOLD:
        return 'Below-prior PIP'

    if sign_prob_pos[j] >= SIGN_STABILITY_THRESHOLD:
        return 'Stable positive'

    if sign_prob_pos[j] <= 1.0 - SIGN_STABILITY_THRESHOLD:
        return 'Stable negative'

    return 'Sign unstable'


classes = [
    sign_class(j)
    for j in order
]

colors = {
    'Stable positive': '#2166ac',
    'Stable negative': '#b2182b',
    'Sign unstable': '#e08214',
    'Below-prior PIP': '0.45'
}

ordered_colors = [
    colors[sign_class(j)]
    for j in order
]

# ------------------------------------------------------------------
# 6. Side-by-side figure
# ------------------------------------------------------------------
fig, (ax_coef, ax_pip) = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(12.8, 0.34 * P + 2.6),
    sharey=True,
    gridspec_kw={
        'width_ratios': [4.8, 1.35],
        'wspace': 0.08
    }
)

# ------------------------------------------------------------------
# 7. Left panel: exact equal-tail posterior intervals
# ------------------------------------------------------------------
for yi, j, cls in zip(
    y_positions,
    order,
    classes
):
    ax_coef.hlines(
        y=yi,
        xmin=COEF_Q025[j],
        xmax=COEF_Q975[j],
        color=colors[cls],
        linewidth=1.5,
        zorder=2
    )

    cap_height = 0.12

    ax_coef.vlines(
        x=[COEF_Q025[j], COEF_Q975[j]],
        ymin=yi - cap_height,
        ymax=yi + cap_height,
        color=colors[cls],
        linewidth=1.1,
        zorder=2
    )

    ax_coef.plot(
        COEF_MEAN[j],
        yi,
        marker='o',
        markersize=5.5,
        color=colors[cls],
        zorder=3
    )

ax_coef.axvline(
    0.0,
    color='0.35',
    linewidth=0.9,
    linestyle='--',
    zorder=1
)

for edge in np.arange(P + 1) - 0.5:
    ax_coef.axhline(
        edge,
        color='0.88',
        linewidth=0.45,
        zorder=0
    )

    ax_pip.axhline(
        edge,
        color='0.88',
        linewidth=0.45,
        zorder=0
    )

if 0 < n_high < P:
    separator_y = n_high - 0.5

    ax_coef.axhline(
        separator_y,
        color='#1f77b4',
        linewidth=1.25,
        zorder=2
    )

    ax_pip.axhline(
        separator_y,
        color='#1f77b4',
        linewidth=1.25,
        zorder=2
    )

ax_coef.set_yticks(
    y_positions
)

ax_coef.set_yticklabels(
    [predictors[j] for j in order],
    fontsize=9
)

ax_coef.set_ylim(
    P - 0.5,
    -0.5
)

ax_coef.set_xlabel(
    f'Unconditional posterior mean and exact '
    f'{100 * CREDIBLE_LEVEL:.0f}% equal-tail credible interval'
)

ax_coef.set_ylabel(
    'Candidate regressor'
)

ax_coef.grid(
    axis='x',
    alpha=0.18
)

# ------------------------------------------------------------------
# 8. Right panel: exact global PIPs
# ------------------------------------------------------------------
ordered_pip = np.asarray(
    PIP,
    dtype=np.float64
)[order]

ax_pip.barh(
    y_positions,
    ordered_pip,
    height=0.72,
    color=ordered_colors,
    edgecolor='none'
)

ax_pip.axvline(
    PIP_HIGH_THRESHOLD,
    color='0.45',
    linewidth=0.8,
    linestyle=':',
    zorder=1
)

ax_pip.set_xlim(
    0.0,
    1.12
)

ax_pip.set_xlabel(
    'PIP (exact)'
)

ax_pip.tick_params(
    axis='y',
    labelleft=False
)

for yi, pip_value, cls in zip(
    y_positions,
    ordered_pip,
    classes
):
    ax_pip.text(
        min(float(pip_value) + 0.015, 1.075),
        yi,
        f'{pip_value:.2f}',
        ha='left',
        va='center',
        fontsize=8,
        color=colors[cls],
        clip_on=False
    )

# ------------------------------------------------------------------
# 9. Section labels
# ------------------------------------------------------------------
coef_xmin, coef_xmax = ax_coef.get_xlim()
coef_span = coef_xmax - coef_xmin

if n_high > 0:
    ax_coef.text(
        coef_xmin - 0.095 * coef_span,
        (n_high - 1) / 2,
        f'Above-prior PIP\n'
        f'(≥ {PIP_HIGH_THRESHOLD:.2f})',
        rotation=90,
        ha='center',
        va='center',
        fontsize=9,
        color='#1f77b4',
        clip_on=False
    )

if n_high < P:
    ax_coef.text(
        coef_xmin - 0.095 * coef_span,
        n_high + (P - n_high - 1) / 2,
        f'Below-prior PIP\n'
        f'(< {PIP_HIGH_THRESHOLD:.2f})',
        rotation=90,
        ha='center',
        va='center',
        fontsize=9,
        color='0.4',
        clip_on=False
    )

# ------------------------------------------------------------------
# 10. Title and legend
# ------------------------------------------------------------------
fig.suptitle(
    'Exact BMA coefficient summary and posterior inclusion probabilities\n'
    'variables grouped relative to the beta-binomial(1,1) prior '
    'inclusion probability of 0.50 and ordered by posterior mean',
    fontsize=13,
    y=0.985
)

legend_handles = [
    Line2D(
        [0], [0],
        marker='o',
        color=colors['Stable positive'],
        markerfacecolor=colors['Stable positive'],
        label='Above-prior PIP, stable positive',
        linestyle='-'
    ),
    Line2D(
        [0], [0],
        marker='o',
        color=colors['Stable negative'],
        markerfacecolor=colors['Stable negative'],
        label='Above-prior PIP, stable negative',
        linestyle='-'
    ),
    Line2D(
        [0], [0],
        marker='o',
        color=colors['Sign unstable'],
        markerfacecolor=colors['Sign unstable'],
        label='Above-prior PIP, sign unstable',
        linestyle='-'
    ),
    Line2D(
        [0], [0],
        marker='o',
        color=colors['Below-prior PIP'],
        markerfacecolor=colors['Below-prior PIP'],
        label='Below-prior PIP',
        linestyle='-'
    )
]

fig.legend(
    handles=legend_handles,
    loc='lower center',
    bbox_to_anchor=(0.5, 0.010),
    ncol=2,
    frameon=False,
    fontsize=9,
    columnspacing=1.4,
    handletextpad=0.6
)

fig.subplots_adjust(
    left=0.20,
    right=0.95,
    top=0.91,
    bottom=0.12
)

# ------------------------------------------------------------------
# 11. Export underlying table
# ------------------------------------------------------------------
coefsummary_tbl = pd.DataFrame({
    'display_order': np.arange(1, P + 1),
    'predictor': [
        predictors[j]
        for j in order
    ],
    'pip': PIP[order],
    'posterior_mean_unconditional': COEF_MEAN[order],
    'posterior_sd_unconditional': COEF_SD[order],
    'posterior_q025_exact': COEF_Q025[order],
    'posterior_median_exact': COEF_Q500[order],
    'posterior_q975_exact': COEF_Q975[order],
    'credible_interval_level': CREDIBLE_LEVEL,
    'credible_interval_excludes_zero': (
        INTERVAL_EXCLUDES_ZERO[order]
    ),
    'p_positive_given_inclusion': sign_prob_pos[order],
    'sign_class': classes,
    'above_prior_pip': (
        PIP[order] >= PIP_HIGH_THRESHOLD
    )
})

save_table(
    coefsummary_tbl,
    'coefsummary_ordered_exact_quantiles'
)

save_fig(
    fig,
    'coefsummary_ordered_exact_quantiles'
)

plt.show()
```

```python
# [PLOT-COEFRIDGELINE] Joy Division–style exact BMA coefficient densities:
# left panel  = conditional-on-inclusion posterior density for each coefficient
#               with a zero-mass marker proportional to (1 - PIP)
# right panel = exact global PIP
#
# Important:
#   - The ridgelines show p(beta_j | beta_j != 0, y), normalized within row
#     for shape comparison only.
#   - The point at zero represents the excluded-model mass (1 - PIP_j).
#   - The right panel shows the exact PIP over all 2^P models.
#
# Ordering:
#   1. predictors with PIP >= prior marginal inclusion probability 0.50
#   2. predictors with PIP < 0.50
# Within each block:
#   descending exact PIP, then descending conditional posterior mean.

from matplotlib.lines import Line2D

PIP_PRIOR_THRESHOLD = 0.50
SIGN_STABILITY_THRESHOLD = 0.95

RIDGE_HEIGHT = 0.82
RIDGE_ALPHA = 0.32
ZERO_MARKER_MIN = 8.0
ZERO_MARKER_MAX = 90.0

SIGN_PROB_COLUMN = 'sign_prob_positive_given_incl_approx'

# ------------------------------------------------------------------
# 1. Recover sign probabilities
# ------------------------------------------------------------------
if SIGN_PROB_COLUMN not in pip_df.columns:
    raise KeyError(
        f'Column {SIGN_PROB_COLUMN!r} not found in pip_df. '
        f'Available columns: {list(pip_df.columns)}'
    )

sign_prob_pos = (
    pip_df
    .set_index('variable')
    .loc[predictors, SIGN_PROB_COLUMN]
    .to_numpy(dtype=np.float64)
)

if sign_prob_pos.shape != (P,):
    raise ValueError(
        f'sign-probability vector has shape {sign_prob_pos.shape}; '
        f'expected ({P},)'
    )

# ------------------------------------------------------------------
# 2. Recover GRID and conditional densities from PASS2
#
# Supports:
#   A) DENS is a dictionary
#   B) GRID and DENS are numerical arrays
# ------------------------------------------------------------------
if 'DENS' not in globals():
    raise NameError(
        '`DENS` is unavailable. Run PASS2 or reload its density output.'
    )

if isinstance(DENS, dict):
    if 'grid' not in DENS:
        raise KeyError("`DENS['grid']` is unavailable")

    if 'conditional_density' not in DENS:
        raise KeyError(
            "`DENS['conditional_density']` is unavailable. "
            f'Available keys: {list(DENS.keys())}'
        )

    coef_grid = np.asarray(
        DENS['grid'],
        dtype=np.float64
    )

    coef_cond_density = np.asarray(
        DENS['conditional_density'],
        dtype=np.float64
    )

else:
    if 'GRID' not in globals():
        raise NameError(
            '`DENS` is an array, so matching numerical `GRID` is required.'
        )

    coef_grid = np.asarray(
        GRID,
        dtype=np.float64
    )

    coef_cond_density = np.asarray(
        DENS,
        dtype=np.float64
    )

print(
    'Coefficient-density arrays:',
    f'grid shape = {coef_grid.shape};',
    f'conditional density shape = {coef_cond_density.shape}'
)

if coef_grid.ndim != 2:
    raise ValueError(
        f'coefficient grid must be two-dimensional; '
        f'got shape {coef_grid.shape}'
    )

if coef_cond_density.ndim != 2:
    raise ValueError(
        f'conditional density must be two-dimensional; '
        f'got shape {coef_cond_density.shape}'
    )

if coef_grid.shape != coef_cond_density.shape:
    raise ValueError(
        f'grid shape {coef_grid.shape} does not match '
        f'density shape {coef_cond_density.shape}'
    )

if coef_grid.shape[0] != P:
    raise ValueError(
        f'density arrays contain {coef_grid.shape[0]} variables, '
        f'but P = {P}'
    )

if not np.isfinite(coef_grid).all():
    raise ValueError('coefficient grid contains non-finite values')

if not np.isfinite(coef_cond_density).all():
    raise ValueError('conditional densities contain non-finite values')

# ------------------------------------------------------------------
# 3. Normalize densities and calculate conditional summaries
# ------------------------------------------------------------------
ridge_density = np.zeros_like(
    coef_cond_density,
    dtype=np.float64
)

conditional_mean = np.zeros(
    P,
    dtype=np.float64
)

conditional_mode = np.zeros(
    P,
    dtype=np.float64
)

for j in range(P):
    grid_j = coef_grid[j]
    dens_j = np.clip(
        coef_cond_density[j],
        0.0,
        np.inf
    )

    area_j = float(
        np.trapezoid(
            dens_j,
            grid_j
        )
    )

    if not np.isfinite(area_j) or area_j <= 0.0:
        raise ValueError(
            f'variable {predictors[j]} has non-positive '
            f'conditional density mass'
        )

    dens_j = dens_j / area_j

    conditional_mean[j] = float(
        np.trapezoid(
            grid_j * dens_j,
            grid_j
        )
    )

    conditional_mode[j] = float(
        grid_j[np.argmax(dens_j)]
    )

    max_density = float(dens_j.max())

    if max_density <= 0.0:
        raise ValueError(
            f'variable {predictors[j]} has zero maximum density'
        )

    # Each row is peak-normalized for shape comparison.
    ridge_density[j] = dens_j / max_density

# ------------------------------------------------------------------
# 4. Sign classification
# ------------------------------------------------------------------
def sign_class(j):
    if PIP[j] < PIP_PRIOR_THRESHOLD:
        return 'Below-prior PIP'

    if sign_prob_pos[j] >= SIGN_STABILITY_THRESHOLD:
        return 'Stable positive'

    if sign_prob_pos[j] <= 1.0 - SIGN_STABILITY_THRESHOLD:
        return 'Stable negative'

    return 'Sign unstable'


colors = {
    'Stable positive': '#2166ac',
    'Stable negative': '#b2182b',
    'Sign unstable': '#e08214',
    'Below-prior PIP': '0.45'
}

# ------------------------------------------------------------------
# 5. Display order
# ------------------------------------------------------------------
high_idx = np.where(
    PIP >= PIP_PRIOR_THRESHOLD
)[0]

low_idx = np.where(
    PIP < PIP_PRIOR_THRESHOLD
)[0]

# Within each block: first exact PIP, then conditional mean.
high_order = high_idx[
    np.lexsort((
        -conditional_mean[high_idx],
        -PIP[high_idx]
    ))
]

low_order = low_idx[
    np.lexsort((
        -conditional_mean[low_idx],
        -PIP[low_idx]
    ))
]

order = np.concatenate([
    high_order,
    low_order
])

n_high = len(high_order)
y_positions = np.arange(P)

classes = [
    sign_class(j)
    for j in order
]

ordered_colors = [
    colors[cls]
    for cls in classes
]

# ------------------------------------------------------------------
# 6. Common coefficient-axis limits
# ------------------------------------------------------------------
grid_min = float(
    np.nanmin(coef_grid)
)

grid_max = float(
    np.nanmax(coef_grid)
)

grid_span = grid_max - grid_min

if grid_span <= 0.0:
    raise ValueError('coefficient grid has zero total span')

x_margin = 0.035 * grid_span
x_left = grid_min - x_margin
x_right = grid_max + x_margin

# ------------------------------------------------------------------
# 7. Side-by-side Joy Division figure
# ------------------------------------------------------------------
fig, (ax_ridge, ax_pip) = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(13.5, 0.38 * P + 2.8),
    sharey=True,
    gridspec_kw={
        'width_ratios': [5.8, 1.35],
        'wspace': 0.08
    }
)

# ------------------------------------------------------------------
# 8. Ridgelines
# ------------------------------------------------------------------
for yi, j, cls in zip(
    y_positions,
    order,
    classes
):
    grid_j = coef_grid[j]
    shape_j = ridge_density[j]

    baseline = float(yi)
    ridge_top = baseline - RIDGE_HEIGHT * shape_j

    # Filled density shape.
    ax_ridge.fill_between(
        grid_j,
        baseline,
        ridge_top,
        color=colors[cls],
        alpha=RIDGE_ALPHA,
        linewidth=0,
        zorder=2
    )

    # Joy Division–style outline.
    ax_ridge.plot(
        grid_j,
        ridge_top,
        color=colors[cls],
        linewidth=1.25,
        zorder=3
    )

    # Conditional posterior mean.
    ax_ridge.plot(
        conditional_mean[j],
        baseline - 0.04,
        marker='o',
        markersize=4.8,
        color=colors[cls],
        zorder=5
    )

    # Point mass at zero:
    # marker area is proportional to excluded-model probability.
    excluded_mass = float(
        np.clip(
            1.0 - PIP[j],
            0.0,
            1.0
        )
    )

    marker_size = (
        ZERO_MARKER_MIN
        + excluded_mass
        * (ZERO_MARKER_MAX - ZERO_MARKER_MIN)
    )

    ax_ridge.scatter(
        [0.0],
        [baseline],
        s=marker_size,
        facecolor='white',
        edgecolor=colors[cls],
        linewidth=1.1,
        zorder=6
    )

# Zero line
ax_ridge.axvline(
    0.0,
    color='0.25',
    linestyle='--',
    linewidth=0.9,
    zorder=1
)

# Shared row guides
for edge in np.arange(P + 1) - 0.5:
    ax_ridge.axhline(
        edge,
        color='0.90',
        linewidth=0.45,
        zorder=0
    )

    ax_pip.axhline(
        edge,
        color='0.90',
        linewidth=0.45,
        zorder=0
    )

# Separator between above- and below-prior PIP blocks
if 0 < n_high < P:
    separator_y = n_high - 0.5

    ax_ridge.axhline(
        separator_y,
        color='#1f77b4',
        linewidth=1.25,
        zorder=7
    )

    ax_pip.axhline(
        separator_y,
        color='#1f77b4',
        linewidth=1.25,
        zorder=7
    )

ax_ridge.set_xlim(
    x_left,
    x_right
)

ax_ridge.set_ylim(
    P - 0.5,
    -1.15
)

ax_ridge.set_yticks(
    y_positions
)

ax_ridge.set_yticklabels(
    [predictors[j] for j in order],
    fontsize=9
)

ax_ridge.set_xlabel(
    'Coefficient value; ridgelines show conditional posterior shape'
)

ax_ridge.set_ylabel(
    'Candidate regressor'
)

ax_ridge.grid(
    axis='x',
    alpha=0.14
)

# ------------------------------------------------------------------
# 9. Exact global PIP panel
# ------------------------------------------------------------------
ordered_pip = np.asarray(
    PIP,
    dtype=np.float64
)[order]

ax_pip.barh(
    y_positions,
    ordered_pip,
    height=0.68,
    color=ordered_colors,
    edgecolor='none'
)

ax_pip.axvline(
    PIP_PRIOR_THRESHOLD,
    color='0.45',
    linestyle=':',
    linewidth=0.9,
    zorder=1
)

ax_pip.set_xlim(
    0.0,
    1.13
)

ax_pip.set_xlabel(
    'PIP (exact)'
)

ax_pip.tick_params(
    axis='y',
    labelleft=False
)

for yi, pip_value, cls in zip(
    y_positions,
    ordered_pip,
    classes
):
    ax_pip.text(
        min(float(pip_value) + 0.015, 1.085),
        yi,
        f'{pip_value:.2f}',
        ha='left',
        va='center',
        fontsize=8,
        color=colors[cls],
        clip_on=False
    )

# ------------------------------------------------------------------
# 10. Section labels
# ------------------------------------------------------------------
ridge_xmin, ridge_xmax = ax_ridge.get_xlim()
ridge_span = ridge_xmax - ridge_xmin

if n_high > 0:
    ax_ridge.text(
        ridge_xmin - 0.075 * ridge_span,
        (n_high - 1) / 2,
        f'Above-prior PIP\n'
        f'(≥ {PIP_PRIOR_THRESHOLD:.2f})',
        rotation=90,
        ha='center',
        va='center',
        fontsize=9,
        color='#1f77b4',
        clip_on=False
    )

if n_high < P:
    ax_ridge.text(
        ridge_xmin - 0.075 * ridge_span,
        n_high + (P - n_high - 1) / 2,
        f'Below-prior PIP\n'
        f'(< {PIP_PRIOR_THRESHOLD:.2f})',
        rotation=90,
        ha='center',
        va='center',
        fontsize=9,
        color='0.4',
        clip_on=False
    )

# ------------------------------------------------------------------
# 11. Title and legend
# ------------------------------------------------------------------
fig.suptitle(
    'Exact BMA coefficient densities and posterior inclusion probabilities\n'
    'ridgelines are conditional on inclusion; circles at zero represent '
    'excluded-model mass',
    fontsize=13,
    y=0.985
)

legend_handles = [
    Line2D(
        [0], [0],
        color=colors['Stable positive'],
        marker='o',
        label='Above-prior PIP, stable positive'
    ),
    Line2D(
        [0], [0],
        color=colors['Stable negative'],
        marker='o',
        label='Above-prior PIP, stable negative'
    ),
    Line2D(
        [0], [0],
        color=colors['Sign unstable'],
        marker='o',
        label='Above-prior PIP, sign unstable'
    ),
    Line2D(
        [0], [0],
        color=colors['Below-prior PIP'],
        marker='o',
        label='Below-prior PIP'
    ),
    Line2D(
        [0], [0],
        color='0.35',
        marker='o',
        markerfacecolor='white',
        label='Circle at zero: size proportional to 1 − PIP'
    )
]

fig.legend(
    handles=legend_handles,
    loc='lower center',
    bbox_to_anchor=(0.5, 0.008),
    ncol=3,
    frameon=False,
    fontsize=8.5,
    columnspacing=1.35,
    handletextpad=0.6
)

fig.subplots_adjust(
    left=0.20,
    right=0.95,
    top=0.91,
    bottom=0.13
)

# ------------------------------------------------------------------
# 12. Export underlying table
# ------------------------------------------------------------------
coefridge_tbl = pd.DataFrame({
    'display_order': np.arange(1, P + 1),
    'predictor': [
        predictors[j]
        for j in order
    ],
    'pip': PIP[order],
    'excluded_model_mass': 1.0 - PIP[order],
    'posterior_mean_unconditional': COEF_MEAN[order],
    'posterior_sd_unconditional': COEF_SD[order],
    'posterior_mean_conditional_on_inclusion': conditional_mean[order],
    'posterior_mode_conditional_on_inclusion': conditional_mode[order],
    'p_positive_given_inclusion': sign_prob_pos[order],
    'sign_class': classes,
    'above_prior_pip': (
        PIP[order] >= PIP_PRIOR_THRESHOLD
    )
})

save_table(
    coefridge_tbl,
    'coefridgeline_ordered'
)

save_fig(
    fig,
    'coefridgeline_ordered'
)

plt.show()
```

```python
# [PLOT-COEFRIDGELINE] Joy Division–style exact BMA coefficient densities:
# left panel  = conditional-on-inclusion posterior density for each coefficient
#               with variable labels placed slightly above the ridge baseline
#               and a zero-mass marker proportional to (1 - PIP)
# right panel = exact global PIP
#
# Important:
#   - The ridgelines show p(beta_j | beta_j != 0, y), normalized within row
#     for shape comparison only.
#   - The point at zero represents the excluded-model mass (1 - PIP_j).
#   - The right panel shows the exact PIP over all 2^P models.
#
# Ordering:
#   1. predictors with PIP >= prior marginal inclusion probability 0.50
#   2. predictors with PIP < 0.50
# Within each block:
#   descending exact PIP, then descending conditional posterior mean.

from matplotlib.lines import Line2D

PIP_PRIOR_THRESHOLD = 0.50
SIGN_STABILITY_THRESHOLD = 0.95

RIDGE_HEIGHT = 0.82
RIDGE_ALPHA = 0.32
ZERO_MARKER_MIN = 8.0
ZERO_MARKER_MAX = 90.0

RIDGE_LABEL_GAP_FRAC = 0.012
RIDGE_VISIBLE_THRESHOLD = 0.015
RIDGE_LABEL_FONTSIZE = 8.5
RIDGE_LABEL_Y_OFFSET = 0.20

SIGN_PROB_COLUMN = 'sign_prob_positive_given_incl_approx'

# ------------------------------------------------------------------
# 1. Recover sign probabilities
# ------------------------------------------------------------------
if SIGN_PROB_COLUMN not in pip_df.columns:
    raise KeyError(
        f'Column {SIGN_PROB_COLUMN!r} not found in pip_df. '
        f'Available columns: {list(pip_df.columns)}'
    )

sign_prob_pos = (
    pip_df
    .set_index('variable')
    .loc[predictors, SIGN_PROB_COLUMN]
    .to_numpy(dtype=np.float64)
)

if sign_prob_pos.shape != (P,):
    raise ValueError(
        f'sign-probability vector has shape {sign_prob_pos.shape}; '
        f'expected ({P},)'
    )

# ------------------------------------------------------------------
# 2. Recover GRID and conditional densities from PASS2
#
# Supports:
#   A) DENS is a dictionary
#   B) GRID and DENS are numerical arrays
# ------------------------------------------------------------------
if 'DENS' not in globals():
    raise NameError(
        '`DENS` is unavailable. Run PASS2 or reload its density output.'
    )

if isinstance(DENS, dict):
    if 'grid' not in DENS:
        raise KeyError("`DENS['grid']` is unavailable")

    if 'conditional_density' not in DENS:
        raise KeyError(
            "`DENS['conditional_density']` is unavailable. "
            f'Available keys: {list(DENS.keys())}'
        )

    coef_grid = np.asarray(
        DENS['grid'],
        dtype=np.float64
    )

    coef_cond_density = np.asarray(
        DENS['conditional_density'],
        dtype=np.float64
    )

else:
    if 'GRID' not in globals():
        raise NameError(
            '`DENS` is an array, so matching numerical `GRID` is required.'
        )

    coef_grid = np.asarray(
        GRID,
        dtype=np.float64
    )

    coef_cond_density = np.asarray(
        DENS,
        dtype=np.float64
    )

print(
    'Coefficient-density arrays:',
    f'grid shape = {coef_grid.shape};',
    f'conditional density shape = {coef_cond_density.shape}'
)

if coef_grid.ndim != 2:
    raise ValueError(
        f'coefficient grid must be two-dimensional; '
        f'got shape {coef_grid.shape}'
    )

if coef_cond_density.ndim != 2:
    raise ValueError(
        f'conditional density must be two-dimensional; '
        f'got shape {coef_cond_density.shape}'
    )

if coef_grid.shape != coef_cond_density.shape:
    raise ValueError(
        f'grid shape {coef_grid.shape} does not match '
        f'density shape {coef_cond_density.shape}'
    )

if coef_grid.shape[0] != P:
    raise ValueError(
        f'density arrays contain {coef_grid.shape[0]} variables, '
        f'but P = {P}'
    )

if not np.isfinite(coef_grid).all():
    raise ValueError('coefficient grid contains non-finite values')

if not np.isfinite(coef_cond_density).all():
    raise ValueError('conditional densities contain non-finite values')

# ------------------------------------------------------------------
# 3. Normalize densities and calculate conditional summaries
# ------------------------------------------------------------------
ridge_density = np.zeros_like(
    coef_cond_density,
    dtype=np.float64
)

conditional_mean = np.zeros(
    P,
    dtype=np.float64
)

conditional_mode = np.zeros(
    P,
    dtype=np.float64
)

for j in range(P):
    grid_j = coef_grid[j]

    dens_j = np.clip(
        coef_cond_density[j],
        0.0,
        np.inf
    )

    area_j = float(
        np.trapezoid(
            dens_j,
            grid_j
        )
    )

    if not np.isfinite(area_j) or area_j <= 0.0:
        raise ValueError(
            f'variable {predictors[j]} has non-positive '
            f'conditional density mass'
        )

    dens_j = dens_j / area_j

    conditional_mean[j] = float(
        np.trapezoid(
            grid_j * dens_j,
            grid_j
        )
    )

    conditional_mode[j] = float(
        grid_j[np.argmax(dens_j)]
    )

    max_density = float(dens_j.max())

    if max_density <= 0.0:
        raise ValueError(
            f'variable {predictors[j]} has zero maximum density'
        )

    ridge_density[j] = dens_j / max_density

# ------------------------------------------------------------------
# 4. Sign classification
# ------------------------------------------------------------------
def sign_class(j):
    if PIP[j] < PIP_PRIOR_THRESHOLD:
        return 'Below-prior PIP'

    if sign_prob_pos[j] >= SIGN_STABILITY_THRESHOLD:
        return 'Stable positive'

    if sign_prob_pos[j] <= 1.0 - SIGN_STABILITY_THRESHOLD:
        return 'Stable negative'

    return 'Sign unstable'


colors = {
    'Stable positive': '#2166ac',
    'Stable negative': '#b2182b',
    'Sign unstable': '#e08214',
    'Below-prior PIP': '0.45'
}

# ------------------------------------------------------------------
# 5. Display order
# ------------------------------------------------------------------
high_idx = np.where(
    PIP >= PIP_PRIOR_THRESHOLD
)[0]

low_idx = np.where(
    PIP < PIP_PRIOR_THRESHOLD
)[0]

high_order = high_idx[
    np.lexsort((
        -conditional_mean[high_idx],
        -PIP[high_idx]
    ))
]

low_order = low_idx[
    np.lexsort((
        -conditional_mean[low_idx],
        -PIP[low_idx]
    ))
]

order = np.concatenate([
    high_order,
    low_order
])

n_high = len(high_order)
y_positions = np.arange(P)

classes = [
    sign_class(j)
    for j in order
]

ordered_colors = [
    colors[cls]
    for cls in classes
]

# ------------------------------------------------------------------
# 6. Common coefficient-axis limits
# ------------------------------------------------------------------
grid_min = float(
    np.nanmin(coef_grid)
)

grid_max = float(
    np.nanmax(coef_grid)
)

grid_span = grid_max - grid_min

if grid_span <= 0.0:
    raise ValueError(
        'coefficient grid has zero total span'
    )

x_margin = 0.035 * grid_span
label_space = 0.10 * grid_span

x_left = grid_min - x_margin - label_space
x_right = grid_max + x_margin

# ------------------------------------------------------------------
# 7. Side-by-side Joy Division figure
# ------------------------------------------------------------------
fig, (ax_ridge, ax_pip) = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(13.5, 0.38 * P + 2.8),
    sharey=True,
    gridspec_kw={
        'width_ratios': [5.8, 1.35],
        'wspace': 0.08
    }
)

# ------------------------------------------------------------------
# 8. Ridgelines and direct variable labels
# ------------------------------------------------------------------
for yi, j, cls in zip(
    y_positions,
    order,
    classes
):
    grid_j = coef_grid[j]
    shape_j = ridge_density[j]

    baseline = float(yi)
    ridge_top = baseline - RIDGE_HEIGHT * shape_j

    visible_idx = np.flatnonzero(
        shape_j >= RIDGE_VISIBLE_THRESHOLD
    )

    if visible_idx.size:
        ridge_start_x = float(
            grid_j[visible_idx[0]]
        )
    else:
        ridge_start_x = float(
            grid_j[0]
        )

    label_x = (
        ridge_start_x
        - RIDGE_LABEL_GAP_FRAC * grid_span
    )

    # Variable label positioned slightly above the ridge baseline.
    ax_ridge.text(
        label_x,
        baseline - RIDGE_LABEL_Y_OFFSET,
        predictors[j],
        ha='right',
        va='center',
        fontsize=RIDGE_LABEL_FONTSIZE,
        color=colors[cls],
        fontweight=(
            'semibold'
            if PIP[j] >= PIP_PRIOR_THRESHOLD
            else 'normal'
        ),
        clip_on=False,
        zorder=8
    )

    ax_ridge.fill_between(
        grid_j,
        baseline,
        ridge_top,
        color=colors[cls],
        alpha=RIDGE_ALPHA,
        linewidth=0,
        zorder=2
    )

    ax_ridge.plot(
        grid_j,
        ridge_top,
        color=colors[cls],
        linewidth=1.25,
        zorder=3
    )

    ax_ridge.plot(
        conditional_mean[j],
        baseline - 0.04,
        marker='o',
        markersize=4.8,
        color=colors[cls],
        zorder=5
    )

    excluded_mass = float(
        np.clip(
            1.0 - PIP[j],
            0.0,
            1.0
        )
    )

    marker_size = (
        ZERO_MARKER_MIN
        + excluded_mass
        * (ZERO_MARKER_MAX - ZERO_MARKER_MIN)
    )

    ax_ridge.scatter(
        [0.0],
        [baseline],
        s=marker_size,
        facecolor='white',
        edgecolor=colors[cls],
        linewidth=1.1,
        zorder=6
    )

# Zero line
ax_ridge.axvline(
    0.0,
    color='0.25',
    linestyle='--',
    linewidth=0.9,
    zorder=1
)

# Shared row guides
for edge in np.arange(P + 1) - 0.5:
    ax_ridge.axhline(
        edge,
        color='0.90',
        linewidth=0.45,
        zorder=0
    )

    ax_pip.axhline(
        edge,
        color='0.90',
        linewidth=0.45,
        zorder=0
    )

# Separator between above- and below-prior PIP blocks
if 0 < n_high < P:
    separator_y = n_high - 0.5

    ax_ridge.axhline(
        separator_y,
        color='#1f77b4',
        linewidth=1.25,
        zorder=7
    )

    ax_pip.axhline(
        separator_y,
        color='#1f77b4',
        linewidth=1.25,
        zorder=7
    )

ax_ridge.set_xlim(
    x_left,
    x_right
)

ax_ridge.set_ylim(
    P - 0.5,
    -1.15
)

ax_ridge.set_yticks(
    y_positions
)

ax_ridge.set_yticklabels(
    []
)

ax_ridge.tick_params(
    axis='y',
    length=0
)

ax_ridge.set_xlabel(
    'Coefficient value; ridgelines show conditional posterior shape'
)

ax_ridge.set_ylabel('')

ax_ridge.grid(
    axis='x',
    alpha=0.14
)

# ------------------------------------------------------------------
# 9. Exact global PIP panel
# ------------------------------------------------------------------
ordered_pip = np.asarray(
    PIP,
    dtype=np.float64
)[order]

ax_pip.barh(
    y_positions,
    ordered_pip,
    height=0.68,
    color=ordered_colors,
    edgecolor='none'
)

ax_pip.axvline(
    PIP_PRIOR_THRESHOLD,
    color='0.45',
    linestyle=':',
    linewidth=0.9,
    zorder=1
)

ax_pip.set_xlim(
    0.0,
    1.13
)

ax_pip.set_xlabel(
    'PIP (exact)'
)

ax_pip.tick_params(
    axis='y',
    labelleft=False
)

for yi, pip_value, cls in zip(
    y_positions,
    ordered_pip,
    classes
):
    ax_pip.text(
        min(float(pip_value) + 0.015, 1.085),
        yi,
        f'{pip_value:.2f}',
        ha='left',
        va='center',
        fontsize=8,
        color=colors[cls],
        clip_on=False
    )

# ------------------------------------------------------------------
# 10. Section labels
# ------------------------------------------------------------------
ridge_xmin, ridge_xmax = ax_ridge.get_xlim()
ridge_span = ridge_xmax - ridge_xmin

if n_high > 0:
    ax_ridge.text(
        ridge_xmin - 0.025 * ridge_span,
        (n_high - 1) / 2,
        f'Above-prior PIP\n'
        f'(≥ {PIP_PRIOR_THRESHOLD:.2f})',
        rotation=90,
        ha='center',
        va='center',
        fontsize=9,
        color='#1f77b4',
        clip_on=False
    )

if n_high < P:
    ax_ridge.text(
        ridge_xmin - 0.025 * ridge_span,
        n_high + (P - n_high - 1) / 2,
        f'Below-prior PIP\n'
        f'(< {PIP_PRIOR_THRESHOLD:.2f})',
        rotation=90,
        ha='center',
        va='center',
        fontsize=9,
        color='0.4',
        clip_on=False
    )

# ------------------------------------------------------------------
# 11. Title and legend
# ------------------------------------------------------------------
fig.suptitle(
    'Exact BMA coefficient densities and posterior inclusion probabilities\n'
    'ridgelines are conditional on inclusion; circles at zero represent '
    'excluded-model mass',
    fontsize=13,
    y=0.985
)

legend_handles = [
    Line2D(
        [0], [0],
        color=colors['Stable positive'],
        marker='o',
        label='Above-prior PIP, stable positive'
    ),
    Line2D(
        [0], [0],
        color=colors['Stable negative'],
        marker='o',
        label='Above-prior PIP, stable negative'
    ),
    Line2D(
        [0], [0],
        color=colors['Sign unstable'],
        marker='o',
        label='Above-prior PIP, sign unstable'
    ),
    Line2D(
        [0], [0],
        color=colors['Below-prior PIP'],
        marker='o',
        label='Below-prior PIP'
    ),
    Line2D(
        [0], [0],
        color='0.35',
        marker='o',
        markerfacecolor='white',
        label='Circle at zero: size proportional to 1 − PIP'
    )
]

fig.legend(
    handles=legend_handles,
    loc='lower center',
    bbox_to_anchor=(0.5, 0.008),
    ncol=3,
    frameon=False,
    fontsize=8.5,
    columnspacing=1.35,
    handletextpad=0.6
)

fig.subplots_adjust(
    left=0.10,
    right=0.95,
    top=0.91,
    bottom=0.13
)

# ------------------------------------------------------------------
# 12. Export underlying table
# ------------------------------------------------------------------
coefridge_tbl = pd.DataFrame({
    'display_order': np.arange(1, P + 1),
    'predictor': [
        predictors[j]
        for j in order
    ],
    'pip': PIP[order],
    'excluded_model_mass': 1.0 - PIP[order],
    'posterior_mean_unconditional': COEF_MEAN[order],
    'posterior_sd_unconditional': COEF_SD[order],
    'posterior_mean_conditional_on_inclusion': conditional_mean[order],
    'posterior_mode_conditional_on_inclusion': conditional_mode[order],
    'p_positive_given_inclusion': sign_prob_pos[order],
    'sign_class': classes,
    'above_prior_pip': (
        PIP[order] >= PIP_PRIOR_THRESHOLD
    )
})

save_table(
    coefridge_tbl,
    'coefridgeline_ordered'
)

save_fig(
    fig,
    'coefridgeline_ordered'
)

plt.show()
```

## Resuming after an interruption

**Same runtime still alive** (kernel restart, cell rerun): run all cells
again with `RUN_FULL_EXACT_P30 = True` — local checkpoints under
`/content` are found automatically and the run continues (idempotent;
finished stages are skipped).

**Runtime destroyed — default free mode** (`USE_GOOGLE_DRIVE = False`):
local checkpoints are gone with the runtime. If you downloaded a
checkpoint bundle beforehand (`download_checkpoint_bundle()`), reopen the
notebook, run the cells through the restore cell, set
`RESTORE_CHECKPOINT_UPLOAD = True`, upload the bundle, then continue with
`RUN_FULL_EXACT_P30 = True`. The bundle is validated (dataset SHA-256,
pinned commit, p, n, priors, dtype, schema) and rejected if incompatible,
so already-processed model ranges are never double-counted.

**Runtime destroyed — optional Drive mode** (`USE_GOOGLE_DRIVE = True`):
reopen and Run all; PASS1/PASS2 resume automatically from the newest
matching Drive checkpoint (at most ~`CHECKPOINT_EVERY_S` of work lost).

In every mode, checkpoints are SHA-256-gated — a checkpoint from different
data, priors, or configuration is rejected instead of silently reused —
and `*.ckpt.npz` files live outside the git clone, are git-ignored, and
are excluded from the results ZIP.
