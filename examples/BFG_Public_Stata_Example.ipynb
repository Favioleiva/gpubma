# BFG applied example: public Grunfeld Stata data

Uses the original Stata Press Grunfeld snapshot in data/public/grunfeld.dta. The documented grunfeld_no_fe baseline is invest on candidates mvalue and kstock, intercept included, no always-in regressors or FE, all 200 rows. No logs, resampling or row deletion. Company/year remain panel identifiers, not additional invented candidates. There are four candidate models, so a fifth ranked model is unavailable. No causal truth, exact global MAP or global probability claims are made.

Fresh Colab runtime; CPU or GPU. Figures reuse the recovered BFG and publication designs. All discovery weights condition on the evaluated set E and are not global PMP/PIP.

## One user configuration cell

**BUDGET is the maximum number of unique candidate models BFG may score.** It maps to the shared `BFGConfig.budget_models` cap. The search may stop earlier, including when the full candidate space has fewer models.

Optional budget choices: `1_000`, `5_000`, `10_000`, `50_000`. They are examples, not an automatically executed ladder.

```python
from pathlib import Path
DATA_PATH = Path('data/public/grunfeld.dta')
TARGET = 'invest'
EXPLANATORY_VARIABLES = ['mvalue', 'kstock']
ALWAYS_IN = []
BUDGET = 5000
SEED = 12345
DEVICE = "auto"
```

## Install and display the runtime

The reviewed release candidate is bundled transparently for validation before publication. It contains the public plotting/reporting modules and the compact archived p30 reference. No hidden helper copy or local source checkout is needed. The alternative branch runs the canonical GitHub pip-install command once these changes are publicly incorporated. Ordinary dependencies may require network installation.

```python
# Temporary reviewed release candidate: the current public main branch
# predates these helpers. After human publication, set USE_REVIEWED_RC=False.
USE_REVIEWED_RC = True
import base64, hashlib, importlib.util, subprocess, sys, tempfile
from pathlib import Path
if any(m == "gpubma" or m.startswith("gpubma.") for m in sys.modules):
    raise RuntimeError("Start a fresh runtime before installing the example package")
requirements = {"numpy":"numpy>=1.26", "pandas":"pandas>=2.1", "scipy":"scipy>=1.11", "torch":"torch>=2.2", "matplotlib":"matplotlib>=3.7", "pyarrow":"pyarrow>=14"}
missing = [r for module,r in requirements.items() if importlib.util.find_spec(module) is None]
if missing:
    subprocess.run([sys.executable,"-m","pip","install","-q",*missing],check=True)
if USE_REVIEWED_RC:
    RC_WHEEL_SHA256 = '8ed81fb49540ae1de02e06de06838b7048b66840aac015af2ae2d5a81b3999c6'
    RC_WHEEL_BASE64 = ''
    raw = base64.b64decode(RC_WHEEL_BASE64, validate=True)
    if hashlib.sha256(raw).hexdigest() != RC_WHEEL_SHA256:
        raise RuntimeError("Release-candidate wheel checksum mismatch")
    runtime_package = Path(tempfile.mkdtemp(prefix="bfg_rc_"))
    wheel_path = runtime_package/'gpubma-0.3.0rc1-py3-none-any.whl'
    wheel_path.write_bytes(raw)
    subprocess.run([sys.executable,"-m","pip","install","--no-deps","--no-index","--target",str(runtime_package/"site"),str(wheel_path)],check=True,capture_output=True,text=True)
    sys.path.insert(0,str(runtime_package/"site"))
else:
    # Canonical public installation command after the reviewed code is published:
    # !pip install -q git+https://github.com/Favioleiva/gpubma.git
    subprocess.run([sys.executable,"-m","pip","install","-q","git+https://github.com/Favioleiva/gpubma.git"],check=True)
import gpubma
from gpubma.bfg.example_workflow import print_environment, ensure_public_data, load_example_data, fit_example, export_example
if USE_REVIEWED_RC:
    assert Path(gpubma.__file__).resolve().is_relative_to((runtime_package/"site").resolve())
environment = print_environment(DEVICE)
```

## Load the dataset without changing the sample

Canonical examples verify the exact repository file SHA-256, fetching from public GitHub only if absent. Generic user data are uploaded when needed. No implicit dropping, imputation, dummy construction or transformation occurs. Always-in controls and candidate columns must be distinct.

```python
from IPython.display import display
import matplotlib.pyplot as plt
get_ipython().run_line_magic("matplotlib", "inline")
DATA_PATH = ensure_public_data('grunfeld', DATA_PATH)
y, X, controls = load_example_data(DATA_PATH, TARGET, EXPLANATORY_VARIABLES, ALWAYS_IN)
print("Observations:", len(y))
```

## Budgeted BFG search

One global hard cap covers wings, genealogy, beam, reconnaissance and elite expansion. No exact enumeration or W-PCS is called.

```python
result, run_metrics = fit_example(y, X, controls, budget=BUDGET, seed=SEED, device=DEVICE, outcome=TARGET)
assert result.n_models_evaluated <= BUDGET
print(result.summary())
```
